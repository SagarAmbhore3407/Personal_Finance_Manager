```python
with open('Samplefile.txt', 'r+') as f1, open('Samplefile2.txt', 'r+') as f2:
    f1lines = f1.readlines()
    
    f2.seek(2)
    for line in f1lines:
        f2.write(line)
        
    f1.seek(0)          # Reset pointer to start before reading
    f1content = f1.read()
    print(f"Samplefile Content:\n{f1content}")
    
    f2.seek(0)          # Reset pointer to start before reading
    f2content = f2.read()
    print(f"Samplefile2 Content:\n{f2content}")
```

```python
# We have 2 nested list, we have to print the student list with second lowest score, 
# and the names should bee in alphabetical order.

students = [['Harry', 37.21], ['Berry', 37.21], ['Tina', 37.2], ['Akriti', 41], ['Harsh', 39]]
students2 = [['Rachel', -50],['Mawer', -50],['Sheen', -50],['Shaheen', 51]]


# Sorting a list according to scores
sorted_lst =sorted(students2, key = lambda x:x[1])

scores = set() # Removes duplicates thats why using set
for item in sorted_lst:
    scores.add(item[1])
        
        
scorlst = sorted(scores)# converted to list of unique sorted values
if len(scorlst)<=2:
    second_low = scorlst[-1]
else:
    second_low = scorlst[1]
            

# Taking only those names which matches the second lowest score.   
names = []
for lst in sorted_lst:
    if lst[1]==second_low:
        names.append(lst[0])

names.sort()   
for name in names:
    print(name)
```

# list:
- List is collection of multiple elements enclosed in square braces [].
- The elements can be heterogenius. list allows us to enclose any type of elements as we want.
- There is no restriction to enclosed specific number of elements, instead the length of list is depends on the our requirement.
- list is Mutable, Means we can update, delete, add and access elements after creation.
- List is ordered in the sequence of how we have created it, so that we can access the element by its index which starts from zero to onwards.
- We can use list() function also to convert any data in list


```python



lst = [1,'himanshu',4.5,1000]

# Accessing directly by using index number
print(lst[1])

# Accessing by loop but no by index
for element in lst:
    print(element)


# Using while loop printing in reverse order
index = len(lst)-1
while index>=0:
    print(lst[index])
    index-=1
    
tup = (2,5,9)
print(tup)  # (2,5,9)
tolist = (list(tup))
print(tolist) # [2,5,9]
    
    
```

```python
# Operations we can perform on list

data = [1,2,3]
print(data)  #[1, 2, 3]

#Adds element at the end in list.
data.append(4)
print(data)  #[1, 2, 3, 4]

# inserts element at the given index
data.insert(1,'a')
print(data)  #[1, 'a', 2, 3, 4]

# Extends the list by given list
data.extend([5,'g'])
print(data)     # [1, 'a', 2, 3, 4, 5, 'g']

# Remove the first occurance of given element
data.remove('a')
print(data)    # [1, 2, 3, 4, 5, 'g']

# take last element and save to another variable.
poped_element = data.pop()
print(poped_element)  # g
print(data)           # [1, 2, 3, 4, 5]

# Reverse the list
data.reverse()
print(data)          # [5, 4, 3, 2, 1]

# Sort the given list in ascending order
data.sort()
print(data)    # [1, 2, 3, 4, 5]

# Returns the index of the first occurance of a given element
index = data.index(3)
print(index)     # 2

# Returns the length of list
length = len(data)
print(length)    #5
```

# Tuple:
- Tuple are also similar to the list.
- but the tuples are immutable, So we cannot modify or update the tuples after creation.
- We cannot perform the modification operations on tuple as we performed on the list.
- tuple is ordered collection of heterogenus elements. Means elements of tuple can be access in by its index order and we can create a tuple which can     contain any type of elements, There is no restriction to use a specific type of elements.
- As tuple is immutable so, it has fewer operations to perform as compared to list.
- We can use tuple() function to convert any datatype into tuple.
  

```python

t = (1,2,3,4,5,3)
print(t) # (1,2,3,4,5,3)

# Count the occurance of the given number
count = t.count(3)
print(count) #2

# Returns the index of the first occurance of a given element
index = t.index(3)
print(index) #2

# # Returns the length of tuple
length = len(t)
print(length) #6

# extend the tuple using the arithmatic operation, But doesnt change original.
# to show use arithmatic operation in print statement.
t+(5,6) 
print(t)   # (1,2,3,4,5,3)
print(t+(5,6)) # # (1,2,3,4,5,3,5,6)


# Similar to +
t*2 
print(t)   # (1,2,3,4,5,3)
print(t*2) # # (1,2,3,4,5,3,1,2,3,4,5,3)

# Checks element in tuple and return Boolean value
print(3 in t) # True

# Sort the tuple but doesnt change original one
print(sorted(t)) # [1,2,3,3,4,5]
print(t)   # # (1,2,3,4,5,3)
```

```python
# using loop
for i in range(len(t)):
    print(t[i])
```

# Set:
- python set is unordered collection of unique elements.
- Sets can have heterogenius elements also but not duplicates, If we tried to create the set with duplictes elements, it will remove all the duplicates of that element.
- As set is mutable so you can add and remove elements but they not allow the duplicates items.
- As they are unordered we cannot access the element from the set.
-  We can convert any datatype to set using the set() function
-  Sets are enclosed in curly braces.
  

  

```python
s1 = {1, 2, 3}
s2 = {2, 3, 4}

# adds elements in set
s1.add(5)   
print(s1)        # {1, 2, 3, 5}

# create one set from both sets and remove duplicates
union_set = s1 | s2
print(union_set)        # {1, 2, 3, 4, 5}

# Returns the elements occurs in in both set.
intersection_set = s1 & s2 
print(intersection_set) # {2, 3}

# Return the with elements in first set but not in both
difference_set = s1 - s2 
print(difference_set)    # {1, 5}

#  Returns a set with elements in either set but not both
symmetric_diff_set = s1 ^ s2   
print(symmetric_diff_set) # {1, 4, 5}
```

# Dictonary:
- python dictonary is a collection of key value pair.
- here key is must be unique, because keys are considered as a index to fetch the perticular value.
- Dictonaries are unordered but maintain the insertion order.
- Dictionary is mutable allows dynamic modification, addition amd removal of key value pairs.
- The value of key is mutable that we can modify that value.
- The dicionary is enclosed in curly braces but the in key value pair and the keys and value is separated by colon.
  

```python

d = {'name':'Sagar','Age':27}
print(d)  # {'name':'Sagar','Age':27}

# Fetching value by get function using key
print(d.get('name')) #Sagar

# Update
d['Age'] = 26
print(d['Age']) # 26

# Adding the pair in dictionary
d['City'] = 'Pune'
print(d)      # {'name':'Sagar','Age':27,'City':'Pune'}

#Returns the list of only keys 
print(d.keys())  # dict_keys(['name', 'Age', 'City'])

# Returns the list of values only.
print(d.values())  # dict_values(['Sagar', 26, 'Pune'])

# Updating the entry using update function.
d.update({'Age':28})
print(d)       # {'name':'Sagar','Age':28,'City':'Pune'}
 
```

```python
# Fetching the values by using loop

for key, value in d.items():
    print(f'{key} : {value}')
```

# String:
- String in python is a sequence of character enclosed in single quotes (''), or Double quotes (""), or sometimes in triple quotes (''' ''' or """ """).
- the string is immutable ponce created the contents cannot be changed.
- We can fetch the character or multiple charactors (part of string) by using the index numbers.

```python
# Function used with string

s = 'Hello World'

# returns the string in lowercase
print(s.lower()) # hello world

# Returns the string in uppercase
print(s.upper()) # HELLO WORLD

# Removes Leading and trailing white spaces for printing not changes the original string.
x= '  Hello World  '
print(len(x))  # 15
y = x.strip()
print(len(y))  # 11

# separates the word and create a list
charLst = s.split()
print(charLst)  # ['Hello','World']

# join the elements from the list using given char
dtlst = ['2025','11','11']
date = '-'.join(dtlst)
print(date) # 2025-11-11

# Replace the specific word with newer one.
newstr = s.replace('Hello','Python')
print(newstr) # Python World

# Returns the starting index of the given word or character.
print(s.find('World')) #6

# Return the length of string
print(len(s)) #11

```

```python
# String Slicing
# While slicing the substring the end index value doesn't print, it will be excluded

str= 'Hello World'
print(str)

# Fecthing a single character using index number
print(str[6]) # W

# Fetching a substring by given the start and end index number.
print(str[0:5]) # Hello

# Slicing by not giving the starting index, so bydefault it will start with the 0th index
print(str[:5]) # Hello

# Similarly we can skip the end point also and bydefault it will print till the end.
print(str[6:]) # World

# The slicing has third parameter also, stepping, eg. step 2,then this will print every second character from string.
print(str[::2]) # HloWrd
print(str[::1]) # Hello World

# This stepping parameter can also change the sequence of the string in reverse order.add()
print(str[::-1]) # dlroW olleH
print(str[::-2]) # drWolH

# We can slice index by giving negative index also
print(str[-5:]) # World
print(str[-6::-1]) # olleH

# Similarly

```

# Ranges:
- In python the ranges data structure typically refers to the range() function.
- It is used in concept of slicing sequences like string, list, etc.
- The ranges doesnt strongly comes under the data structure section, but as it creates the data and structure it according to our requirement we can say that it is used as the data structure.
- range() function can have 3 parameters similar to string slicing ranges function also work as a slicing.

```python
a = range(1,10)
print(a)  # range(1,10)

for i in a:
    print(f'{i} ',end='') # 1 2 3 4 5 6 7 8 9

print(f'\n{a[3]}')  # 4

# This will print the odd nums from 1 to 10
for j in range(1,10,2):
    print(f'{j} ',end="") # 1 3 5 7 9

print(f'\n{list(a)}')   #[1, 2, 3, 4, 5, 6, 7, 8, 9]

lst = list(a)

for g in range(0,len(lst)+5,3):
    if g<len(lst):
        print(f'{lst[g]}, ',end='')
    else :    
        print(f'lst[{g}]=NA, ',end='') # 1, 4, 7, lst[9]=NA, lst[12]=NA,


```

# Frozensets:
- Frozensets are similar to the normal set but unlike sets the frozensets cannot be modified after creation.
- We cannot delete, update, or remove the data from the frozen set.

```python
mylst = [3,4,1,3]
fset = frozenset(mylst)

print(fset)  # frozenset({1, 3, 4})

#fset.add('cherry')  # Attribute error
#fset.remove(3)   # Attribute error

fset2 = frozenset([2,5,6,2,3,4])
print(fset2)      #frozenset({2, 3, 4, 5, 6})

print(fset.union(fset2))  # frozenset({1, 2, 3, 4, 5, 6})
print(fset.intersection(fset2))  # frozenset({3, 4})


```

# LinkedList
- A linkedlist is linear data structure composed of nodes, where each node contains data and refference to the next data point which creates proper sequence of nodes.
- Linked list is similar to array, In array we have to specify the datatype and the size of array before creation, and array also needs only a contagious memory location to store the array element in sequence in memory. 
- But in linkedlist we doesnt need to specify the size and datatype and the main advantage is, The linked list can be stored in the fragmented memory also, as it uses the refference node to locate the next element.

```python
class LinkedList:
    def __init__(self, data=None, next=None):
        self.data = data
        self.next = None
        
    def setdata(self,data):
        self.data = data
        
    def getdata(self):
        return self.data
    
    def setnext(self,next):
        self.next = next
        
    def getnext(self):
        return self.next
    

head = LinkedList(1)
node2 = LinkedList(2)
node3 = LinkedList(3)
node4 = LinkedList(4)

head.setnext(node2)
node2.setnext(node3)
node3.setnext(node4)

print(head.getdata(),end='->')
print(node2.getdata(),end='->')
print(node3.getdata(),end='->')
print(node4.getdata(),end='->')  # 1->2->3->4->
print()


# this function will traverse the list and print one by one element.
def traverse(head):
    if not head:
        print(None)
        return
    print(head.getdata(),end='->')
    traverse(head.getnext())
    
traverse(head)           # 1->2->3->4->None

# this function will return the length of Linkedlist
def lengthLL(head):
    if not head:
        return 0
    return 1+lengthLL(head.getnext())

print(lengthLL(head))    # 4


        
# this function will append the element at the last of Linkedlist 
def append(head,data=None,):
    if not head.getnext():
        head.setnext(data)
        return
    append(head.getnext(),data)
 
# appending the new node   
newAppend = LinkedList(5)
append(head,newAppend)
 
traverse(head)        # 1->2->3->4->5->None
print(lengthLL(head))  # 5


# This function is used to insert an node at starting, middle or at end.
def insertnode(head, new, k):
    if k == 1:
        new.setnext(head)
        return new
    
    if head is not None:
        head.setnext(insertnode(head.getnext(),new,k-1))
    return head    

new = LinkedList(12)
a = insertnode(head,new,6)

traverse(a)

```
